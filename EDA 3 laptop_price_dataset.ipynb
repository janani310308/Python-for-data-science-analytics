```python
#Price Distribution
plt.figure(figsize=(8,5))
sns.histplot(df['price'],kde=True)
plt.title("Laptop Price Distribution")
plt.show()
```

```python
#price by brand
plt.figure(figsize=(10,5))
sns.boxplot(x='brand',y='price',data=df)
plt.title("Laptop Price By Brand")
plt.xticks(rotation=45)
plt.show()
```

```python
#correction analysis
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
#cpu impact on price
plt.figure(figsize=(10,6))
sns.boxplot(x="cpu",y="price",data=df)
plt.title("CPU Impact on Laptop Price")
plt.xlabel("CPU")
plt.ylabel("Price")
plt.show()
```

```python
#ram significantly increase price
df.groupby("ram")["price"].mean()
sns.scatterplot(x="ram",y="price",data=df)
plt.title("RAM vs Price")
plt.show()
```

```python
#expensive laptop
brand_price=df.groupby("brand")["price"].max().sort_values(ascending=False)
print(brand_price.head())
#visualization
brand_price.head(10).plot(kind='bar')
plt.title("Top Expensive Laptop Brands")
plt.ylabel("price")
plt.show()
```

```python
df.groupby("category")["price"].mean()
sns.boxplot(x="category",y="price",data=df)
plt.title("Gaming vs Student Laptop")
plt.show()
```

```python
df.groupby("cpu")["price"].mean()
sns.boxplot(x="cpu",y="price",data=df)
plt.title("CPU Type vs Price")
plt.show()
```

```python
#highest average price

top_cpu=df.groupby("cpu")["price"].mean().sort_values(ascending=False).head(10)
top_cpu.plot(kind='bar')
plt.title("CPU vs Average Price")
plt.ylabel("Price")
plt.show()
```

## Key Insights

#### Insight 1

Laptop prices vary significantly across brands, with premium brands like Apple, Dell, and high-end HP models generally priced higher than others.

#### Insight 2

 Gaming laptops are more expensive on average compared to student and home laptops due to better processors, higher RAM, and dedicated graphics cards.

#### Insight 3

There is a strong positive relationship between RAM size and price — laptops with 16GB or more RAM tend to be significantly costlier.

#### Insight 4

 SSD storage has a major impact on price. Devices with larger SSD (512GB or 1TB) are priced higher than those with smaller storage.

#### Insight 5

 Laptops with dedicated graphics cards (Nvidia/AMD) are more expensive than those with integrated graphics, especially in gaming categories.

#### Insight 6

Laptops with dedicated graphics cards (Nvidia/AMD) are more expensive than those with integrated graphics, especially in gaming categories.

#### Insight 7

 Processor type strongly influences price — Intel i7 / Ryzen 7 laptops are more expensive than i3 / Ryzen 3 models.

#### Insight 8

 Battery capacity shows a mild positive correlation with price, as higher battery (Wh) is usually found in premium laptops.

#### Insight 9

 Weight does not significantly impact price, indicating portability is not the primary pricing factor.

#### Insight 10

 Windows 11 laptops dominate the dataset, but OS type has minimal direct impact on pricing.

####

## CONCLUSION

The exploratory data analysis of the laptop price dataset reveals that pricing is primarily driven by hardware specifications and usage category. High-performance components such as advanced processors (Intel i7/Ryzen 7), larger RAM, and dedicated graphics cards significantly increase laptop prices.

Gaming laptops emerge as the most expensive segment due to their superior configurations, while student and home laptops are comparatively more affordable. Storage type and size, especially SSD capacity, also play a crucial role in determining price.

Although factors like screen size and battery capacity have some influence, they are less impactful compared to core hardware features. Additionally, weight and operating system show minimal effect on pricing trends.

Overall, the analysis indicates that performance-oriented features are the key determinants of laptop pricing, and consumers pay a premium for better speed, graphics capability, and multitasking efficiency.
